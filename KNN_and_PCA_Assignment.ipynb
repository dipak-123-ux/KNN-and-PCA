1.	What is K-Nearest Neighbors (KNN) and how does it work in both classification and regression problems?
    -	K-Nearest Neighbors (KNN) is a supervised, non-parametric, and lazy machine learning algorithm that classifies or predicts the target of a new data point based on the proximity and values of its closest training neighbors. It does not build an explicit mathematical model during training; instead, it memorizes the entire training dataset and performs computations only when a prediction is requested.
How KNN Works: Core Steps
Choose K: Select the number of nearest neighbors (K) to examine, such as 3 or 5.
Calculate Distance: Measure the distance between the new data point and all stored training points using a distance metric.
Find Neighbors: Sort the distances in ascending order and pick the top K closest data points
Make Prediction: Apply an aggregation rule based on whether the problem is classification or regression.

Classification Problem Mechanics
In classification tasks, the Elastic kNN Guide notes that the algorithm tallies the class labels of the chosen K nearest neighbors. The new data point is assigned to the class that appears most frequently (the majority vote). For instance, if K = 5 and 4 neighbors are red circles while 1 neighbor is a blue square, the model classifies the new point as a red circle. Odd values for K are typically chosen in binary classification to prevent voting ties.
Regression Problem Mechanics
In regression tasks, the algorithm ignores class voting and instead extracts the numerical target values from the K nearest neighbors. It calculates the simple average or mean of these numerical values to establish the prediction for the new query point. For example, if predicting a house price with K = 3 and the nearest neighboring houses cost $300,000, $310,000, and $320,000, the predicted price is the average: $310,000

2.	What is the Curse of Dimensionality and how does it affect KNN performance?
    -	The Curse of Dimensionality refers to a phenomenon where analyzing and organizing data becomes exponentially more difficult as the number of features (dimensions) increases. In high-dimensional spaces, data points become isolated from each other, rendering the concept of "closeness" or proximity meaningless.
How It Affects KNN Performance
Because KNN relies entirely on calculating distances to identify neighbors, it is one of the algorithms hardest hit by this phenomenon in the following ways:
Distance Convergence: As the number of dimensions increases, the distance between the nearest neighbor and the farthest neighbor approaches the same value. Because every point becomes roughly equidistant from every other point, KNN can no longer reliably distinguish which neighbors are actually "close."
Data Sparsity: To maintain the same density of data points and ensure a stable neighborhood, the amount of training data required must grow exponentially with each added feature. Without an unmanageable amount of data, the model effectively calculates distances through mostly empty space.
  Irrelevant Features: If you add columns or features that have no predictive power, KNN treats them with equal weight during standard distance calculations (like Euclidean distance). This injects random noise into the distance metric, drowning out the signal from meaningful features.
    Computational Slowdown: Since KNN is a "lazy learner" that calculates distances on the         fly during inference, adding dozens or hundreds of dimensions dramatically increases the mathematical operations required for every single prediction, causing severe latency.

3.	What is Principal Component Analysis (PCA)? How is it different from feature selection?
    -	Principal Component Analysis (PCA) is an unsupervised dimensionality reduction technique that transforms a large set of correlated variables into a smaller set of uncorrelated variables called principal components.
It does this by projecting the data onto new orthogonal (perpendicular) axes that capture the maximum possible variance in the dataset.
Core Difference: PCA vs. Feature Selection
While both methods aim to reduce the size of your dataset to improve model performance and combat the curse of dimensionality, they use entirely opposite philosophies:
Metric / Attribute	Principal Component Analysis (PCA)	Feature Selection
Strategy	Feature Extraction: Combines existing features into entirely new ones.	Feature Retention: Selects a subset of existing features without altering them.
Data Alteration	Modifies the original data; creates a new coordinate system.	Leaves original values intact; simply drops unwanted columns.
Interpretability	Low: The new components are mathematical mixtures (e.g., 0.5 × Age + 0.2 × Income).	High: The remaining variables keep their original meaning and business context.
Information Loss	Minimal; aims to retain as much variance (information) as possible.	Higher risk; completely discards all information in the dropped columns.

4.	What are eigenvalues and eigenvectors in PCA, and why are they important?
    - Eigenvalues and eigenvectors are the core mathematical engines of Principal Component Analysis (PCA) that dictate how the original high-dimensional data is rotated and compressed. When PCA analyzes a dataset, it computes a covariance matrix to see how the features vary together. Eigenvectors and eigenvalues are the factors extracted from this matrix to reshape the data.
Why Are They Important in PCA
Without eigenvalues and eigenvectors, PCA would not be able to identify which directions in your data carry the most valuable signal. They serve three critical purposes:
Establishing the New Coordinate System
The eigenvectors calculated from the covariance matrix are completely orthogonal (at a 90∘ angle) to one another. This ensures that the newly created principal components are completely uncorrelated, eliminating the issue of multicollinearity that often plagues datasets before PCA is applied.
Ranking Feature Importance
Eigenvalues act as a mathematical leaderboard for your data. A larger eigenvalue means its corresponding eigenvector captures more of the dataset's total variance. By sorting the eigenvalues in descending order, PCA automatically ranks the principal components from most informative to least informative.
Deciding How Many Dimensions to Keep
Eigenvalues allow you to calculate the Explained Variance Ratio to determine exactly how many dimensions can be dropped without losing critical information.

5.	How do KNN and PCA complement each other when applied in a single pipeline?
    -	KNN and PCA complement each other by forming a powerful pipeline where PCA fixes the structural flaws of high-dimensional data, allowing KNN to calculate distances quickly and accurately. While KNN is exceptionally good at finding local patterns based on proximity, it fails when there are too many features. PCA solves this by restructuring the feature space before KNN looks at it.
Step-by-Step Pipeline Mechanics
To implement this successfully, the algorithms must follow a strict, ordered pipeline:
Scale the Data: Standardize features (Mean = 0, Variance = 1) so large-scale features don't distort PCA's variance calculations or KNN's distance metrics.
Apply PCA: Transform the scaled data into principal components. Decide how many components to retain based on the cumulative explained variance (typically aiming for 85% to 95%).
Train KNN: Feed the lower-dimensional PCA projections into the KNN algorithm for final classification or regression.
Direct Comparison: Before vs. After the Pipeline
Metric / Scenario	KNN Alone (High-Dim Data)	PCA + KNN Pipeline
Distance Computations	Slow; calculates across many noisy axes.	Fast; calculates across a few optimized axes.
Handling Correlated Features	Poor; overweights duplicate or redundant signals.	Excellent; PCA pre-decorrelates all features.
Risk of Overfitting	High; easily distracted by random feature noise.	Low; noise is filtered out in lower eigenvalues.
    



```python
#Q6. Train a KNN Classifier on the Wine dataset with and without feature
#scaling. Compare model accuracy in both cases.
#(Include your Python code
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Load data and split (70% train, 30% test)
wine = load_wine()
X_train, X_test, y_train, y_test = train_test_split(wine.data, wine.target, test_size=0.3, random_state=42)

# Without Scaling
knn_unscaled = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
acc_unscaled = accuracy_score(y_test, knn_unscaled.predict(X_test))

# With Scaling
scaler = StandardScaler()
X_train_scaled, X_test_scaled = scaler.fit_transform(X_train), scaler.transform(X_test)
knn_scaled = KNeighborsClassifier(n_neighbors=5).fit(X_train_scaled, y_train)
acc_scaled = accuracy_score(y_test, knn_scaled.predict(X_test_scaled))

print(f"Without scaling: {acc_unscaled * 100:.2f}% | With scaling: {acc_scaled * 100:.2f}%")
```

```python
#Q7. Train a PCA model on the Wine dataset and print the explained variance
#ratio of each principal component.
#(Include your Python code
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Load the dataset
wine = load_wine()
X = wine.data

# 2. Standardize the features (Crucial step for PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Fit the PCA model
pca = PCA()
pca.fit(X_scaled)

# 4. Extract and print the explained variance ratios
explained_variance = pca.explained_variance_ratio_

print("Explained Variance Ratio per Principal Component:\n")
for i, variance in enumerate(explained_variance, 1):
    print(f"PC{i:02d}: {variance * 100:.2f}%")

print(f"\nTotal variance captured by first 2 components: {sum(explained_variance[:2]) * 100:.2f}%")
```

```python
#Q8. Train a KNN Classifier on the PCA-transformed dataset (retain top 2
#components). Compare the accuracy with the original dataset.
#(Include your Python code
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Load data and perform a 70-30 train-test split
wine = load_wine()
X_train, X_test, y_train, y_test = train_test_split(wine.data, wine.target, test_size=0.3, random_state=42)

# 2. Standardize features (mandatory baseline for both models)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. Baseline: Train KNN on the original scaled dataset (all 13 features)
knn_orig = KNeighborsClassifier(n_neighbors=5).fit(X_train_scaled, y_train)
acc_orig = accuracy_score(y_test, knn_orig.predict(X_test_scaled))

# 4. Dimensionality Reduction: Reduce dataset to top 2 principal components
pca = PCA(n_components=2)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

# 5. PCA Model: Train KNN on the 2D PCA-transformed features
knn_pca = KNeighborsClassifier(n_neighbors=5).fit(X_train_pca, y_train)
acc_pca = accuracy_score(y_test, knn_pca.predict(X_test_pca))

print(f"Accuracy on Original Scaled Dataset: {acc_orig * 100:.2f}%")
print(f"Accuracy on PCA-Transformed Dataset (2 PCs): {acc_pca * 100:.2f}%")
```

```python
#Q9. Train a KNN Classifier with different distance metrics (euclidean,
#manhattan) on the scaled Wine dataset and compare the results.
#(Include your Python code
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Load data and split (70% train, 30% test)
wine = load_wine()
X_train, X_test, y_train, y_test = train_test_split(wine.data, wine.target, test_size=0.3, random_state=42)

# 2. Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. Train and evaluate with Euclidean distance
knn_euclidean = KNeighborsClassifier(n_neighbors=5, metric='euclidean').fit(X_train_scaled, y_train)
acc_euclidean = accuracy_score(y_test, knn_euclidean.predict(X_test_scaled))

# 4. Train and evaluate with Manhattan distance
knn_manhattan = KNeighborsClassifier(n_neighbors=5, metric='manhattan').fit(X_train_scaled, y_train)
acc_manhattan = accuracy_score(y_test, knn_manhattan.predict(X_test_scaled))

print(f"KNN with Euclidean Distance Accuracy: {acc_euclidean * 100:.2f}%")
print(f"KNN with Manhattan Distance Accuracy: {acc_manhattan * 100:.2f}%")
```

```python
#Q10. You are working with a high-dimensional gene expression dataset to
#classify patients with different types of cancer.
#Due to the large number of features and a small number of samples, traditional models
#overfit.
#Explain how you would:
# Use PCA to reduce dimensionality
# Decide how many components to keep
# Use KNN for classification post-dimensionality reduction
# Evaluate the model
# Justify this pipeline to your stakeholders as a robust solution for real-world
#biomedical data
#(Include your Python code
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, f1_score, make_scorer

# 1. Re-simulate high-dimensional biomedical data
X_raw, y_raw = make_classification(
    n_samples=100, n_features=5000, n_informative=50,
    n_classes=3, n_clusters_per_class=1, random_state=42
)

# 2. Determine optimal components (90% variance threshold)
scaler_temp = StandardScaler()
X_temp = scaler_temp.fit_transform(X_raw)
pca_temp = PCA().fit(X_temp)
cumulative_variance = np.cumsum(pca_temp.explained_variance_ratio_)
n_components_90 = np.argmax(cumulative_variance >= 0.90) + 1

# 3. Re-build the explicit pipeline
biomedical_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=n_components_90)),
    ('knn', KNeighborsClassifier(n_neighbors=5, metric='euclidean'))
])

# 4. FIXED: Convert metric strings into explicit scorer objects using make_scorer
scoring_metrics = {
    'accuracy': make_scorer(accuracy_score),
    'f1_macro': make_scorer(f1_score, average='macro')  # Handles multi-class data explicitly
}

# 5. Run Stratified Cross-Validation safely
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_results = cross_validate(biomedical_pipeline, X_raw, y_raw, cv=cv, scoring=scoring_metrics)

print("--- 5-Fold Cross-Validation Performance ---")
print(f"Mean Accuracy: {np.mean(cv_results['test_accuracy']) * 100:.2f}% (+/- {np.std(cv_results['test_accuracy']) * 100:.2f}%)")
print(f"Mean F1-Score: {np.mean(cv_results['test_f1_macro']) * 100:.2f}%")
```
